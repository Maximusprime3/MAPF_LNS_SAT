```python
log_sources = [
        ("LNS-SAT", Path("lns_clean/logs/LNS")),
        ("MDD-SAT", Path("lns_clean/logs/WholeSolve")),
        ]

# First create the plot
ax = plot_solver_time_cdf(
        log_sources,
        title="Overall Performance",
)

# Override the legend labels to show only the solver names (remove all numbers)
handles, labels = ax.get_legend_handles_labels()
new_labels = [label.split(" (")[0] for label in labels]  # Keep only the part before " ("
ax.legend(handles, new_labels)

# Set size and save
ax.figure.set_size_inches(4, 3)
```

```python
# Set the figure size first
ax.figure.set_size_inches(3, 2)
ax.set_xlabel("Elapsed Time [s]")
ax.set_ylabel("Completed Experiments")
# Then save
ax.figure.savefig("overall_performance.svg", 
                  format='svg', 
                  bbox_inches='tight')
```

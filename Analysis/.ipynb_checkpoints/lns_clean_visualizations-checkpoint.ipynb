# LNS Clean Solver Log Visualizations

This notebook turns the CSV traces in `lns_clean/logs` into publication-ready figures. It concentrates on the two solver variants that are available in the repository:

* **LNS SAT MDD solve** — the incremental LNS pipeline stored under `lns_clean/logs/LNS` (plus the harder `HardRooms` cases).
* **SAT MDD solve** — the end-to-end solver runs stored under `lns_clean/logs/WholeSolve`.

Each visualization cell reads directly from the CSVs, so you can refresh the figures after running new experiments without editing any code. The helper utilities also standardise column names and derived metrics (e.g., seconds instead of milliseconds) so downstream plots stay tidy.


## Environment preparation

```python
from pathlib import Path
import math

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams.update({"figure.dpi": 120, "axes.spines.top": False, "axes.spines.right": False})

DATA_ROOT = Path("../lns_clean/logs").resolve()
if not DATA_ROOT.exists():
    raise FileNotFoundError(f"Expected to find {DATA_ROOT}. Adjust DATA_ROOT if you moved the notebook.")

DATASET_LABELS = {
    "LNS": "LNS SAT MDD solve",
    "WholeSolve": "SAT MDD solve",
}

available_dirs = sorted(path.name for path in DATA_ROOT.iterdir() if path.is_dir())
print(f"Loading logs from: {DATA_ROOT}")
print("Available log groups:", ", ".join(available_dirs))
```

```python
def extract_stem(path_like):
    if pd.isna(path_like):
        return None
    try:
        return Path(str(path_like)).stem
    except Exception:
        return str(path_like)


def coerce_columns(df, columns):
    for col in columns:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")


def add_time_columns(df, ms_columns):
    for col in ms_columns:
        if col in df.columns:
            seconds_col = col.replace("_ms", "_s")
            df[seconds_col] = df[col] / 1000.0


def load_table(table_name, data_root=DATA_ROOT):
    frames = []
    for dataset_dir in data_root.iterdir():
        if not dataset_dir.is_dir():
            continue
        dataset_name = dataset_dir.name
        if dataset_name.startswith("."):
            continue
        if dataset_name == "HardRooms":
            csv_path = dataset_dir / f"{table_name}.csv"
            if csv_path.exists():
                df = pd.read_csv(csv_path)
                if not df.empty:
                    df["dataset_origin"] = dataset_name
                    df["map_folder"] = dataset_dir.name
                    frames.append(df)
            continue
        for map_dir in dataset_dir.iterdir():
            if not map_dir.is_dir():
                continue
            csv_path = map_dir / f"{table_name}.csv"
            if csv_path.exists():
                df = pd.read_csv(csv_path)
                if not df.empty:
                    df["dataset_origin"] = dataset_name
                    df["map_folder"] = map_dir.name
                    frames.append(df)
    if not frames:
        return pd.DataFrame()
    combined = pd.concat(frames, ignore_index=True)
    return combined


def attach_experiment_metadata(df, experiments_df):
    if df.empty:
        return df
    meta_cols = [
        "experiment_id",
        "dataset",
        "dataset_origin",
        "dataset_pretty",
        "map_folder",
        "map_name",
        "scenario_name",
        "num_agents",
        "is_solved",
        "total_runtime_s",
        "total_mdd_build_s",
        "total_cnf_build_s",
        "total_lazy_solver_wall_s",
    ]
    available_meta = [col for col in meta_cols if col in experiments_df.columns]
    return df.merge(experiments_df[available_meta], on="experiment_id", how="left")
```

```python
# Load experiments summary and derive consistent metadata columns
experiments = load_table("experiments")
if experiments.empty:
    raise RuntimeError("No experiment rows found under lns_clean/logs. Double-check the path or regenerate the logs.")

EXPERIMENT_NUMERIC = [
    "num_agents",
    "solved",
    "makespan_success",
    "total_runtime_ms",
    "total_cnf_clauses",
    "total_cnf_variables",
    "total_mdd_build_ms",
    "total_cnf_build_ms",
    "total_lazy_wall_ms",
    "total_lazy_solver_wall_ms",
    "total_lazy_solver_reported_ms",
]
coerce_columns(experiments, EXPERIMENT_NUMERIC)

experiments["map_name"] = experiments.get("map_path", pd.Series([None] * len(experiments))).apply(extract_stem)
experiments["scenario_name"] = experiments.get("scenario_path", pd.Series([None] * len(experiments))).apply(extract_stem)
experiments["dataset_origin"] = experiments["dataset_origin"].fillna("LNS")
experiments["dataset"] = experiments["dataset_origin"].replace({"HardRooms": "LNS"})
experiments["dataset_pretty"] = experiments["dataset"].map(DATASET_LABELS).fillna(experiments["dataset"])
experiments["solved"] = experiments["solved"].astype("Int64")
experiments["makespan_success"] = experiments["makespan_success"].astype("Int64")
experiments["is_solved"] = experiments["solved"] == 1

TIME_COLUMNS_EXP = [
    "total_runtime_ms",
    "total_mdd_build_ms",
    "total_cnf_build_ms",
    "total_lazy_wall_ms",
    "total_lazy_solver_wall_ms",
    "total_lazy_solver_reported_ms",
]
add_time_columns(experiments, TIME_COLUMNS_EXP)

experiments["cnf_clause_density"] = experiments["total_cnf_clauses"] / experiments["total_cnf_variables"].replace({0: np.nan})
experiments.head()
```

```python
experiment_summary = (
    experiments.groupby(["dataset_pretty", "map_name"], dropna=False)
    .agg(
        runs=("experiment_id", "count"),
        solved_runs=("is_solved", "sum"),
        median_runtime_s=("total_runtime_s", "median"),
        p90_runtime_s=("total_runtime_s", lambda x: x.quantile(0.9)),
        max_agents=("num_agents", "max"),
    )
    .reset_index()
)
experiment_summary.sort_values(["dataset_pretty", "map_name"]).head(20)
```

```python
# Load the detailed iteration-level tables and attach experiment metadata
raw_tables = {}
for table_name in ["makespan_attempts", "local_zones", "lazy_iterations", "waiting_attempts"]:
    df = load_table(table_name)
    if df.empty:
        print(f"⚠️ No rows found for {table_name}.")
    else:
        raw_tables[table_name] = df

MAKESPAN_NUMERIC = [
    "makespan_attempt",
    "makespan",
    "zones_attempted",
    "zones_solved",
    "total_waiting_attempts",
    "total_lazy_iterations",
    "total_cnf_clauses",
    "total_cnf_variables",
    "total_mdd_build_ms",
    "total_cnf_build_ms",
    "total_lazy_wall_ms",
    "total_lazy_solver_wall_ms",
    "total_lazy_solver_reported_ms",
    "attempt_wall_ms",
    "solved",
]
LOCAL_ZONE_NUMERIC = [
    "makespan_attempt",
    "zone_attempt",
    "zone_positions",
    "zone_fraction",
    "conflicts",
    "agents",
    "start_t",
    "end_t",
    "waiting_attempts",
    "total_lazy_iterations",
    "total_cnf_clauses",
    "total_cnf_variables",
    "total_mdd_build_ms",
    "total_cnf_build_ms",
    "total_lazy_wall_ms",
    "total_lazy_solver_wall_ms",
    "total_lazy_solver_reported_ms",
    "solved",
]
LAZY_NUMERIC = [
    "makespan_attempt",
    "zone_attempt",
    "waiting_attempt",
    "iteration",
    "clauses_before",
    "variables",
    "clauses_added",
    "total_clauses_after",
    "new_vertex_collisions",
    "new_edge_collisions",
    "total_vertex_collisions",
    "total_edge_collisions",
    "iteration_wall_ms",
    "solver_wall_ms",
    "solver_reported_ms",
    "used_assumptions",
    "reset_solver",
    "satisfiable",
    "num_decisions",
    "num_propagations",
]
WAITING_NUMERIC = [
    "makespan_attempt",
    "zone_attempt",
    "waiting_attempt",
    "waiting_budget",
    "zone_positions",
    "segment_count",
    "agent_count",
    "start_t",
    "end_t",
    "cnf_clauses",
    "cnf_variables",
    "mdd_build_ms",
    "cnf_build_ms",
    "attempt_wall_ms",
    "lazy_iterations",
    "lazy_total_wall_ms",
    "lazy_solver_wall_ms",
    "lazy_solver_reported_ms",
    "applied_waiting",
    "extended_window",
    "solved",
]

makespan_attempts = raw_tables.get("makespan_attempts", pd.DataFrame())
if not makespan_attempts.empty:
    coerce_columns(makespan_attempts, MAKESPAN_NUMERIC)
    add_time_columns(
        makespan_attempts,
        [
            "total_mdd_build_ms",
            "total_cnf_build_ms",
            "total_lazy_wall_ms",
            "total_lazy_solver_wall_ms",
            "total_lazy_solver_reported_ms",
            "attempt_wall_ms",
        ],
    )
    makespan_attempts["solved"] = makespan_attempts["solved"].astype("Int64")
    makespan_attempts = attach_experiment_metadata(makespan_attempts, experiments)

local_zones = raw_tables.get("local_zones", pd.DataFrame())
if not local_zones.empty:
    coerce_columns(local_zones, LOCAL_ZONE_NUMERIC)
    add_time_columns(
        local_zones,
        [
            "total_mdd_build_ms",
            "total_cnf_build_ms",
            "total_lazy_wall_ms",
            "total_lazy_solver_wall_ms",
            "total_lazy_solver_reported_ms",
        ],
    )
    local_zones["solved"] = local_zones["solved"].astype("Int64")
    local_zones = attach_experiment_metadata(local_zones, experiments)

lazy_iterations = raw_tables.get("lazy_iterations", pd.DataFrame())
if not lazy_iterations.empty:
    coerce_columns(lazy_iterations, LAZY_NUMERIC)
    add_time_columns(
        lazy_iterations,
        ["iteration_wall_ms", "solver_wall_ms", "solver_reported_ms"],
    )
    lazy_iterations = attach_experiment_metadata(lazy_iterations, experiments)

waiting_attempts = raw_tables.get("waiting_attempts", pd.DataFrame())
if not waiting_attempts.empty:
    coerce_columns(waiting_attempts, WAITING_NUMERIC)
    add_time_columns(
        waiting_attempts,
        [
            "mdd_build_ms",
            "cnf_build_ms",
            "attempt_wall_ms",
            "lazy_total_wall_ms",
            "lazy_solver_wall_ms",
            "lazy_solver_reported_ms",
        ],
    )
    waiting_attempts["solved"] = waiting_attempts["solved"].astype("Int64")
    waiting_attempts = attach_experiment_metadata(waiting_attempts, experiments)

{
    "makespan_attempts": len(makespan_attempts),
    "local_zones": len(local_zones),
    "lazy_iterations": len(lazy_iterations),
    "waiting_attempts": len(waiting_attempts),
}
```

## 1. Performance skyline (CNF size vs runtime)

This plot shows all experiment runs coloured by map, with the Pareto-optimal frontier highlighted for each solver family. It helps identify which maps dominate in terms of CNF size and runtime trade-offs.


```python
def compute_pareto_front(df, x_col, y_col, group_cols):
    if df.empty:
        return df.assign(is_pareto=False)
    data = df.dropna(subset=[x_col, y_col]).copy()
    data["is_pareto"] = False
    for keys, group in data.groupby(group_cols):
        group = group.sort_values([x_col, y_col], ascending=[True, True])
        best_y = math.inf
        selected = []
        for idx, row in group.iterrows():
            if row[y_col] <= best_y:
                selected.append(idx)
                best_y = row[y_col]
        data.loc[selected, "is_pareto"] = True
    return data

skyline_data = compute_pareto_front(
    experiments,
    x_col="total_cnf_variables",
    y_col="total_runtime_s",
    group_cols=["dataset_pretty", "map_name"],
)

if skyline_data.empty:
    print("No experiment data available for the skyline plot.")
else:
    datasets = skyline_data["dataset_pretty"].unique()
    n_cols = len(datasets)
    fig, axes = plt.subplots(1, n_cols, figsize=(7 * n_cols, 6), sharey=False, sharex=False)
    if n_cols == 1:
        axes = [axes]

    for ax, dataset_name in zip(axes, datasets):
        subset = skyline_data[skyline_data["dataset_pretty"] == dataset_name]
        if subset.empty:
            ax.set_visible(False)
            continue
        sns.scatterplot(
            data=subset,
            x="total_cnf_variables",
            y="total_runtime_s",
            hue="map_name",
            style="is_solved",
            ax=ax,
            alpha=0.65,
        )
        pareto_subset = subset[subset["is_pareto"]].sort_values("total_cnf_variables")
        for map_name, group in pareto_subset.groupby("map_name"):
            ax.plot(
                group["total_cnf_variables"],
                group["total_runtime_s"],
                linewidth=2.5,
                label=f"{map_name} Pareto",
            )
        ax.set_title(f"{dataset_name} — CNF vs runtime skyline")
        ax.set_xlabel("Total CNF variables")
        ax.set_ylabel("Runtime (s)")
        ax.set_xscale("log")
        ax.set_yscale("log")
        ax.legend(loc="best", fontsize=9)
    plt.tight_layout()
```

```python
from pathlib import Path
from load_experiment_data import (
    ActualMapAgentKey,
    load_and_split_by_actual_map,
)

log_root = Path("../lns_clean/logs/LNS")
map_frames, actual_map_agent_frames = load_and_split_by_actual_map(log_root)

# Now you can access data by actual map name and agent count
# For example, to get empty-16-16 with 10 agents:
key = ActualMapAgentKey(actual_map_name="empty-16-16", num_agents=10)
data = actual_map_agent_frames[key]
```

```python
map_names= ['empty-16-16', 'empty-32-32', 'empty-48-48', 
            'room-32-32-4', 'room-64-64-16', 'room-64-64-8',
            'Berlin_1_256','Paris_1_256', 
            'warehouse-10-20-10-2-2', 'warehouse-20-40-10-2-2']
number_of_agents = [10, 20, 50, 100, 200]
```

```python
for map_name in map_names:
    for agent_number in number_of_agents:
        key = ActualMapAgentKey(actual_map_name=map_name, num_agents=agent_number)
        data = actual_map_agent_frames[key]
        print(data['actual_map_name'][0])
        print(data['num_agents'][0])
        if map_name == 'empty-16-16' and agent_number == 10:
            print(data.keys())
```
